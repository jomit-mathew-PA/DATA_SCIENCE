```python
# importing the module called pandas to peerform the requiored operations
import pandas as pd
```

```python
# reading the file using pandas method
df=pd.read_csv("car_prices.csv")
```

```python
# checking the null values in dataframe
df.isna().sum()
```

```python
# Displaying the column names
df.columns
```

```python
# Here i want to know how many rows and columns does my dataframe consist
df.shape
```

```python
# Since there null values present in my dataset i am trying to fill the columns using the respected values
transmission_mode=df['transmission'].mode()
df['transmission'].fillna(transmission_mode[0],inplace=True)
df['transmission'].isna().sum()
```

```python
# Here we are filling the null values in condition column using mean because it is numerical column
condition_mean=df['condition'].mean()
df['condition'].fillna(condition_mean,inplace=True)
df['condition'].isna().sum()
```

```python
# filling null values using mode because it is a categorical column
body_mode=df['body'].mode()
df['body'].fillna(body_mode[0],inplace=True)
df['body'].isna().sum()
```

```python
# filling null values in model coulmn using mode
model_mode=df['model'].mode()
df['model'].fillna(model_mode[0],inplace=True)
df['model'].isna().sum()
```

```python
# fill null values in make column using mode
make_mode=df['make'].mode()
df['make'].fillna(make_mode[0],inplace=True)
df['make'].isna().sum()
```

```python
trim_mode=df['trim'].mode()
df['trim'].fillna(trim_mode[0],inplace=True)
df['trim'].isna().sum()
```

```python
# filling vin column using mode
vin_mode=df['vin'].mode()
df['vin'].fillna(vin_mode[0],inplace=True)
df['vin'].isna().sum()
```

```python
# filling nan values in odometer column using mean
odometer_mean=df['odometer'].mean()
df['odometer'].fillna(odometer_mean,inplace=True)
df['odometer'].isna().sum()
```

```python
# filling null values in color column using mode
color_mode=df['color'].mode()
df['color'].fillna(color_mode[0],inplace=True)
df['color'].isna().sum()
```

```python
# filling nan values using mode
interior_mode=df['interior'].mode()
df['interior'].fillna(interior_mode[0],inplace=True)
df['interior'].isna().sum()
```

```python
# filling nan value using mean
mmr_mean=df['mmr'].mean()
df['mmr'].fillna(mmr_mean,inplace=True)
df['mmr'].isna().sum()
```

```python
# filling nan values using mean
selling_mean=df['sellingprice'].mean()
df['sellingprice'].fillna(selling_mean,inplace=True)
df['sellingprice'].isna().sum()
```

```python
# here we are droping the null values in sale date column using dropna
df.dropna(subset='saledate',axis=0,inplace=True)
df['saledate'].isna().sum()
```

```python
# now let us check how many null values are present in our data frame
# after checking the null values if there no null values present then our data frame it is ready for further analysis
# now we will name our new data frame
CLEANED_DF=df
```
